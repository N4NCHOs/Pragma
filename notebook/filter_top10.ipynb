```python
import re
```

```python
TARGET_ASSETS = {
    "BITCOIN", "BTC",
    "ETHEREUM", "ETH",
    "TETHER", "USDT",
    "BNB",
    "XRP",
    "USD COIN", "USDC",
    "SOLANA", "SOL",
    "TRON", "TRX",
    "DOGECOIN", "DOGE",
    "HYPERLIQUID", "HYPE"
}
```

```python
NAMES_PATTERN = re.compile(
    r'\b(bitcoin|ethereum|tether|bnb|xrp|usd coin|solana|tron|dogecoin|hyperliquid)\b', 
    re.IGNORECASE
)
TICKERS_PATTERN = re.compile(
    r'\b(BTC|ETH|USDT|USDC|SOL|TRX|DOGE|HYPE)\b'
)
```

```python
def is_top_10_news(news_item):
    """
    Filter using Regex to check for Top 10 assets.
    """
    title = news_item.get("title", "")
    body = news_item.get("description", "")
    full_text = f"{title}. {body}"
    
     # Check if any full name is mentioned (case-insensitive)
    if NAMES_PATTERN.search(full_text):
        return True
        
    # Check if any exact ticker is mentioned (case-sensitive)
    if TICKERS_PATTERN.search(full_text):
        return True
        
    return False
```

```python
# Simulated JSON payload from the ingestion batching tool
incoming_news_batch = [
    {
        "id": "1",
        "title": "Hyperliquid Launches New L2 Order Book",
        "description": "The HYPE token surged today following the release of the new trading engine.",
        "url": "https://example.com/news/1"
    },
    {
        "id": "2",
        "title": "Random Meme Coin reaches 1 million Market Cap",
        "description": "PepeElonInu is trending on decentralized exchanges.",
        "url": "https://example.com/news/2"
    },
    {
        "id": "3",
        "title": "Is the Solana network congested again?",
        "description": "Validators are reporting issues processing SOL transactions this morning.",
        "url": "https://example.com/news/3"
    },
    {
        "id": "4",
        "title": "Sun hype reaches new levels in Miami",
        "description": "Tourists are flocking to the beaches. The sun is out, and the hype is real.",
        "url": "https://example.com/news/4" 
        # Note: 'hype' and 'sun'
    }
]
```

```python
filtered_news = []
for news in incoming_news_batch:
    if is_top_10_news(news):
        filtered_news.append(news)
        print(f"[PASSED] {news['title']}")
    else:
        print(f"[DROPPED] {news['title']}")
```
